# scalar arithmetic(mathemetical operation on single value)

```python
import numpy as np

arr=np.array([1,2,3,4,5])
print(arr +1)
print(arr -1)
print(arr *2)
print(arr /4)
print(arr **3)
```

# vector arithmetic (1D array)

```python
array3=np.array([1,2,3,4,5,6])
array4=np.array([1.0,2.3,5.1,8.6])
print(np.sqrt(array3))
print(np.round(array4))
print(np.floor(array4))
print(np.pi)
```

# 2 dimension array(nested array) OR matrix multiplicatio

```python
import numpy as np

A=np.array([[1,2],
           [3,4]])
B=np.array([[5,6],
           [7,8]])

print(A.shape)#no of rows and columns
print(A*B)
```

# slicing

```python
import numpy as np
lst=[1,2,3,4,5,6,7,8,9]
result=np.array(lst)
print("1.result[1:8]-> element from index 1 to 7 :",result[1:8])
print("2.result[1:]-> element from index 1 to end :",result[1:])
print("3.result[:1]-> element from index 0 :",result[:1])
print("4.result[::-1]-> reverse the lst :",result[::-1])


```

#  negative slicing

```python
lst=[1,2,3,4,5,6,7]
arr=np.array(lst)
print("5.arr[-3:-1]-> from 3rd last to 2rd last element:",arr[-3:-1])
print("6.arr[::3]-> every 3rd element :",arr[::3])
```

# slicing 2D array

```python
import numpy as np

nested_list1=[
    [1,2,3,5,9,8],
    [4,5,6,7,8,9]

]

new_arr=np.array(nested_list1)
print(new_arr[0, 1:4])
print(new_arr[0:2,1])
print(new_arr[0:2,1:4])
```
