# Aggregate Functions in NumPy
# ===============================
### Aggregate functions are used to summarize data.
### They usually take an array as input and return a single value (or summary).
### Examples: sum, mean, max, min, argmin, argmax, etc.

```python
import numpy as np
arr=np.array([
    [2,4,6,8,10],
    [1,3,5,7,9]
])

print("sum of all elements",np.sum(arr))

print("mean(average)",np.mean(arr))

print("maximum value",np.max(arr))

print("minimum value",np.min(arr))

print("index of minimum value",np.argmin(arr))

print("index of maximum value",np.argmax(arr))

#sum across columns (axis=0)
print("column-wise sum",np.sum(arr,axis=0))

#sum across rows (axis=1)
print("row-wise sum",np.sum(arr,axis=1))

# product of all elements
print("product of all elements",np.prod(arr))

# count of array
print(arr.size)

# how many to create  a random array
rand_arr=np.random.randint(1,100,size=39)
print(rand_arr)
```

# 1. Basic Filtering + Sum
# Given an array of marks:

# → Filter only marks greater than 50
# → Find the sum of these filtered marks

```python
import numpy as np
marks = np.array([45, 78, 23, 90, 67, 88, 34, 59])
mark50=marks[marks>50]
arr1=np.sum(mark50)
print(arr1)
```

# 2. Filtering + Mean
# Given an array:


# → Select values between 40 and 90 (exclusive)
# → Calculate their mean

```python
arr = np.array([12, 45, 67, 89, 34, 22, 90, 100])
marks=arr[(arr>40) &(arr<90)]
arr1=np.mean(marks)
print(arr1)
```

# 5. Filter Before Sum
# Create a 2D array:
array = np.array([[10, 20, 30],
                  [40, 50, 60],

# → Filter elements greater than 30 and less than 80
# → Compute the sum of these filtered elements

```python
array = np.array([[10, 20, 30],
                  [40, 50, 60],
                   [70, 80, 90]])
marks=array[(array>30) & (array<80)]
arr1=np.sum(marks)
print("greater than 30 and less than 80",marks)
print("sum of these filtered elements",arr1)
```

# 7. Chained Conditions + Aggregation
'''Create a random array of 15 integers between 1 and 100'''
# → Filter numbers either less than 20 OR greater than 80
# → Find their mean and sum


```python
import numpy as np
arr1=np.random.randint(1,100,size=15)
num1=arr1[(arr1<20)|(arr1>80)]
sum2=np.sum(num1)
mean2=np.mean(num1)

print("either less than 20 OR greater than 80",num1)
print("sum",sum2)
print("mean",mean2)
```


# 8. Aggregation on Condition
# Given a 2D array:
matrix = np.array([[23, 45, 67],
                   [89, 12, 34],
                   [56, 78, 90]])
# → Filter all even numbers
# → Calculate their maximum and minimum

```python
matrix = np.array([[23, 45, 67],
                   [89, 12, 34],
                   [56, 78, 90]])
even_number=matrix[matrix%2==0]
n1=np.min(even_number)
n2=np.max(even_number)

print("even number",even_number)
print("calculate minimum",n1)
print("calculate maximum",n2)
```

# 9. Filter Outside Range
# Given an array:
arr = np.array([12, 45, 60, 75, 30, 95, 100, 20])
# → Filter all values outside the range 40–90 (<40 or >90)
# → Find the sum and count of these filtered elements

```python
arr = np.array([12, 45, 60, 75, 30, 95, 100, 20])
mark=arr[(arr<40) |(arr>90)]
count=len(mark)
n1=np.sum(mark)
print("(<40 or >90)",mark)
print("count of these filtered elements",count)
print("sum",n1)
```

# 10. Filter with Logical NOT (~)
# Given:
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
# → Filter all elements NOT greater than 40
# → Compute the mean of these remaining elements

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
mark=arr[~(arr>40)]
n1=np.mean(mark)
print("Filter all elements NOT greater than 40",mark)
print("Compute the mean of these remaining elements",n1)
```

# 11. Create a 5×5 array of random integers between 1 and 100
# → Filter elements that are multiples of 5
# → Find their sum, mean, and total count

```python
import numpy as np
arr=np.random.randint(1,100,size=(5,5))
mul5=arr[arr%5==0]
n1=np.sum(mul5)
n2=np.mean(mul5)
total_count=len(mul5)

print(" Create a 5×5 array of random integers between 1 and 100",arr)
print("Filter elements that are multiples of 5",mul5)
print("sum",n1)
print("mean",n2)
print("total count",total_count)
```
