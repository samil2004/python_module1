# Aggregate Functions in NumPy
# ===============================
### Aggregate functions are used to summarize data.
### They usually take an array as input and return a single value (or summary).
### Examples: sum, mean, max, min, argmin, argmax, etc.

```python
import numpy as np
arr=np.array([
    [2,4,6,8,10],
    [1,3,5,7,9]
])

print("sum of all elements",np.sum(arr))

print("mean(average)",np.mean(arr))

print("maximum value",np.max(arr))

print("minimum value",np.min(arr))

print("index of minimum value",np.argmin(arr))

print("index of maximum value",np.argmax(arr))

#sum across columns (axis=0)
print("column-wise sum",np.sum(arr,axis=0))

#sum across rows (axis=1)
print("row-wise sum",np.sum(arr,axis=1))

# product of all elements
print("product of all elements",np.prod(arr))

# count of array
print(arr.size)

# how many to create  a random array
rand_arr=np.random.randint(1,100,size=39)
print(rand_arr)
```

### reshape

```python
import numpy as np
arr=np.array([1,2,3,4,5,6])
reshaped=arr.reshape(2,3)
reshaped1=arr.reshape(3,2)
print("reshaped():\n",reshaped)
print("reshaped():\n",reshaped1)
```

### flatten() -> convert multi dimensional array into 1D
### ravel()


```python
import numpy as np
arr=np.array([
    [2,4,6,8,10],
    [1,3,5,7,9]
])
flat=arr.flatten()
print("flatten():",flat)

# ravel()
flat1=arr.ravel()
print("ravel():",flat1)
```

### transpose

```python
import numpy as np
arr=np.array([
    [2,4,6,8,10],
    [1,3,5,7,9]
])
transposed=arr.transpose()
print("transpose():\n",transposed)
```

### vstack and hstack

```python
import numpy as np
A=np.array([[1,2,3]])
B=np.array([[4,5,6]])
vertical=np.vstack((A,B))
horizontal=np.hstack((A,B))

print("verical():\n:",vertical)
print("horizontal():",horizontal)
```

### split():- split an array into multiple part

```python
import numpy as np
arr=np.array([10,20,30,40,50,60])
parts=np.split(arr,2)
print("split():\n",parts)
```

### unique():-> find unique elements in a array

```python
import numpy as np
a=np.array([1,2,3,4,5,3,4,5])
und=np.unique(a)
print("unique():",und)
```
